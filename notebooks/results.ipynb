```python
import pandas as pd
import matplotlib.pyplot as plt

# Your data
df = pd.DataFrame({
    "specific_error": [
        "Adjacent Key", "Missing Complete Thought", "Missing Letter",
        "Missing Object", "Missing Subject", "Repeating Letter", "Scrambled Letter"
    ],
    "accuracy": [
        0.5098039215686274, 0.4409090909090909, 0.43636363636363634,
        0.4354066985645933, 0.441747572815534, 0.512987012987013,
        0.40963855421686746
    ]
})

# Define grouping rule: indices 0, 5, 6 = word-level errors
word_level_idx = [0, 5, 6]
df["error_category"] = ["word-level" if i in word_level_idx else "sentence-level"
                        for i in range(len(df))]

# Assign colors
colors = df["error_category"].map({
    "word-level": "skyblue",
    "sentence-level": "orange"
})

# Plot
plt.figure(figsize=(10, 5))
plt.bar(df["specific_error"], df["accuracy"], color=colors)
plt.xticks(rotation=45, ha='right')

plt.ylabel("Accuracy")
plt.title("Accuracy by Error Type — Model_Journal_BertTuned35")

# Legend
plt.legend(handles=[
    plt.Rectangle((0,0),1,1,color="skyblue", label="Word-level Errors"),
    plt.Rectangle((0,0),1,1,color="orange", label="Sentence-level Errors")
])

plt.tight_layout()
plt.show()
```
